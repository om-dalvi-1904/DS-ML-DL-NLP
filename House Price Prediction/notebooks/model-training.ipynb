```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
df = pd.read_csv('IndianHouses.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df = df.dropna()
```

```python
df.shape
```

```python
df = df.drop(['Locality', 'Per_Sqft'], axis=1)
```

```python
df.head()
```

```python
print(df['Furnishing'].value_counts())
print(df['Status'].value_counts())
print(df['Transaction'].value_counts())
print(df['Type'].value_counts())
```

```python
df.head()
```

```python
df['Furnishing'] = df['Furnishing'].str.strip().str.title().replace({
    'Furnished': 1,
    'Unfurnished': 2,
    'Semi-Furnished': 3
})
```

```python
df['Status'] = df['Status'].str.strip().str.title().replace({
    'Ready_To_Move': 1,
    'Almost_Ready': 2,
})
```

```python
df['Transaction'] = df['Transaction'].str.strip().str.title().replace({
    'Resale': 1,
    'New_Property': 2,
})
```

```python
df['Type'] = df['Type'].str.strip().str.title().replace({
    'Builder_Floor': 1,
    'Apartment': 2,
})
```

```python
df.head()
```

```python
df.info()
```

```python
# dependant and independant feature

x = df[['Area', 'BHK', 'Bathroom', 'Furnishing', 'Parking', 'Status', 'Transaction', 'Type']]
y = df['Price']
```

```python
x.head()
```

```python
y.head()
```

```python
# train test split

from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.25, random_state = 42)
```

```python
# train the model

from sklearn.linear_model import LinearRegression
```

```python
linear = LinearRegression(n_jobs=-1)
```

```python
linear.fit(x_train, y_train)
```

```python
y_pred = linear.predict(x_test)
```

```python
# performance metrics

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
print(mse)
print(mae)
print(rmse)
print(r2)
```
