## Decision Tree Classifier (Iris Dataset)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
iris
```

```python
iris['DESCR']
```

```python
# dependant and independant features

x = pd.DataFrame(iris.data, columns=['sepal length', 'sepal width', 'petal length', 'petal width'])
y = iris['target']
```

```python
# train test split

from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.20, random_state=42)
```

```python
# train the model

from sklearn.tree import DecisionTreeClassifier
```

```python
tree_classifier = DecisionTreeClassifier()
```

```python
tree_classifier.fit(x_train, y_train)
```

```python
# visualize the tree

from sklearn import tree
```

```python
plt.figure(figsize=(15, 10))
tree.plot_tree(tree_classifier, filled=True, feature_names=x.columns, class_names=iris.target_names)
```

```python
y_pred = tree_classifier.predict(x_test)
```

```python
# post pruning the tree

tree_classifier_pruned = DecisionTreeClassifier(max_depth=2)
tree_classifier_pruned.fit(x_train, y_train)
tree.plot_tree(tree_classifier_pruned, filled=True, feature_names=x.columns, class_names=iris.target_names)
```

```python
# compare the accuracy scores

from sklearn.metrics import classification_report
```

```python
print("Classification Report: \n", classification_report(y_test, y_pred))
print("Classification Report (Pruned): \n", classification_report(y_test, tree_classifier_pruned.predict(x_test)))
```
