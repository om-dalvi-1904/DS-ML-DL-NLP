```python
from nltk.tokenize import sent_tokenize
```

```python
corpus = '''Hello everyone, my name is Om Dalvi.
I'm pursuing B.Tech from PCCOE. Learning new technologies is my passion.
'''
```

```python
# sentence tokenization

sent_tokenize(corpus)
```

```python
documents = sent_tokenize(corpus)
```

```python
for sentence in documents:
    print(sentence)
```

```python
# word tokenization. para to words, sentence to words

from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
# workPunct tokenize tokenizes punctuation marks as well

from nltk.tokenize import WordPunctTokenizer
```

```python
word_tokenizer = WordPunctTokenizer()
```

```python
word_tokenizer.tokenize(corpus)
```

```python
# treeBankWordTokenizer tokenizes only words.

from nltk.tokenize import TreebankWordTokenizer
```

```python
tree_word_tokenizer = TreebankWordTokenizer()
```

```python
tree_word_tokenizer.tokenize(corpus)
```
